```python
import os
import numpy as np
import joblib
from sklearn.neighbors import KNeighborsClassifier
```

```python
embedding_dir = "/home/debashis/SIH'26/Inter Clg/Attendance-System/data/trained_embeddings"
model_path = "/home/debashis/SIH'26/Inter Clg/Attendance-System/models/knn_model.pkl"
```

```python
#load embedding
X = []
y = []

for filename in sorted(os.listdir(embedding_dir)):
    if not filename.endswith(".npy"):
        continue

    roll = os.path.splitext(filename)[0]

    file_path  = os.path.join(
        embedding_dir,
        filename
    )

    embeddings = np.load(file_path)
    print(f"Roll {roll}: " f"{embeddings.shape}")

    #add embeddings
    for  embedding in embeddings:
        X.append(embedding)
        y.append(roll)
        
```

```python
#convert to numpy
X = np.array(X, dtype=np.float32)
y = np.array(y)

print("\nDataset INFO-")
print(f"Total embeddings : {len(X)}")
print(f"Embedding shape  : {X.shape}")
print(f"Classes          : {len(np.unique(y))}")
print(f"Roll numbers     : {np.unique(y)}")
```

```python
#create KNN
knn = KNeighborsClassifier(
    n_neighbors = 5,
    metric = "cosine",
    weights = "distance"
)

#train model
knn.fit(X, y)
print("Training colpleted")
```

```python
#save model
joblib.dump(
    knn,
    model_path
)

print(f"\nModel saved to: {model_path}")
```
